```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a dummy DataFrame
df = pd.DataFrame({'x': [1, 2, 3], 'y': [4, 5, 6]})
print('Dummy DataFrame:')
print(df)

# Plot the data
plt.plot(df['x'], df['y'])
plt.title('Dummy Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.show()

print('Hello from test.ipynb!')
```
