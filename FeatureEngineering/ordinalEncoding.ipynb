```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['Delhi', 'pune', 'Tokyo']])
```

```python
import pandas as pd
df= pd.DataFrame({
    'city': ['Tokyo', 'Delhi', 'Tokyo', 'pune', 'Delhi'],})
df.head()
```

```python
encoder.fit_transform(df[['city']])
```
