## Target Guided Ordinal encoding

It is a technique used to encode categorical variables based on thier relationship with the 
target variable. This encoding technique is useful when we have a categorical variable with 
a large number of unique categiries and we want to use this variable as a feature in our machine
learning model.

In target Guided Ordinal Encoding , we replace each category in the categorical variable with a numerical value based on the median or meann of the target 
variable for that category. This creates a monotonic relationship between the categorical variable and the targert variable which can improve the predictive power of the model. 

```python
import pandas as pd
df=pd.DataFrame({"cities":["New York","London","Paris","Tokyo","London","Paris"],
                 "price":[100,200,300,400,500,600],
                 })
```

```python
df
```

```python
mean_price=df.groupby("cities")["price"].mean().to_dict()
mean_price
```

```python
df["city_encoded"]=df["cities"].map(mean_price)
df
```
