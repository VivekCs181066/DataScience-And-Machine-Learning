```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
```

```python
X, Y = make_regression(n_samples=100, n_features=1, n_informative=1,n_targets=1, noise=20)
```

```python
plt.scatter(X,Y)
```

```python
lr.fit(X,Y)
```

```python
np.mean(cross_val_score(lr,X,Y,scoring="r2",cv=10))
```

```python
class GDRegerssor:

    def __init__(self, learning_rate, n_iters):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.m=34.41
        self.b=-120
    def fit(self, X, Y):
        n_samples, n_features = X.shape

        for _ in range(self.n_iters):
            loss_slope=-2 * np.sum(Y - (self.m * X.ravel() + self.b))
            step_size = self.lr * loss_slope
            self.b=self.b-step_size
            print(loss_slope,self.b)
        print(self.b)
```

```python
class GDRegerssorWithMandB:

    def __init__(self, learning_rate, n_iters):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.m=100
        self.b=-120
    def fit(self, X, Y):
        n_samples, n_features = X.shape

        for _ in range(self.n_iters):
            loss_slope_b=-2 * np.sum(Y - (self.m * X.ravel() + self.b))
            loss_slope_m=-2 * np.sum((Y - (self.m * X.ravel() + self.b))*X.ravel())
            step_size_b = self.lr * loss_slope_b
            step_size_m = self.lr * loss_slope_m
            self.b=self.b-step_size_b
            self.m=self.m-step_size_m
            # print(loss_slope,self.b)
        print(self.m,self.b)
    def predict(self,X):
        return self.m * X + self.b
```

```python
gd_with_m_and_b=GDRegerssorWithMandB(learning_rate=0.001, n_iters=60)
```

```python
gd_with_m_and_b.fit(X,Y)
# (np.float64(-1.4685108455750129), array([81.01264475]))
```
