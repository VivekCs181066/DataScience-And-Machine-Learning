8. skiprows parameter

```python
pd.read_csv("aug_train.csv",skiprows=[1,4])
```

```python
# Skip rows using a function
def skip_rows(index):
    # Skip rows 1 and 4 (0-based indexing)
    if index in [1,4]:
        return True
    return False

df = pd.read_csv("aug_train.csv", skiprows=skip_rows)
```

12. Coverter

```python
def rename(name):
    if name=='Royal Challengers Bangalore':
        return 'RCB'
    elif name=='Delhi Daredevils':
        return 'DD'
    elif name=='Delhi Capitals':
        return 'DC'
    elif name=='Mumbai Indians':
        return 'MI'
    return name
pd.read_csv("IPL Matches 2008-2020.csv",converters={'team1':rename,'team2':rename,'toss_winner':rename,'winner':rename}).head()
```
