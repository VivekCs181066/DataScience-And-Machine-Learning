```python
import pandas as pd
```

```python
import random
```

```python
l=[]
for i in range(1000):
    ## Rolling one dice only
    l.append(random.randint(1,6))
print(l)
```

```python
ll=[]
for i in range(1000):
    ## Rolling 2 dice at same time
    a=random.randint(1,6)
    b=random.randint(1,6)
    ll.append(a+b)
print(ll)
```

```python
seriesList=(pd.Series(l).value_counts()/pd.Series(l).count().sum()).sort_index()
seriesListList=(pd.Series(ll).value_counts()/pd.Series(ll).count().sum()).sort_index()
```

```python
seriesList
seriesListList
```

```python
import numpy as np
np.cumsum(seriesListList)
```

```python
np.cumsum(seriesListList).plot(kind='bar')
```

```python
seriesList.plot(kind='bar')
seriesListList.plot(kind='bar')
```
