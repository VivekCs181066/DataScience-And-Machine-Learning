```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from numpy.random import normal 
```

```python
sample=normal(loc=50, scale=5, size=1000)
```

```python
sample
```

```python
## Plot histogram to understand the distribution of data
```

```python
plt.hist(sample, bins=10)
```

```python
## calculate the sample mean and standaard deviation
sample_mean=sample.mean()
sample_std=sample.std()
```

```python
## Fit the distribution with the above parameters
from scipy.stats import norm
```

```python
distribution=norm(loc=sample_mean, scale=sample_std)
```

```python
values=np.linspace(min(sample), max(sample), 100)
```

```python
probabilities=[ distribution.pdf(value) for value in values]
```

```python
##  plot the histogram and pdf
```

```python
plt.hist(sample, bins=10)
```

```python
plt.hist(sample, bins=10,density=True)
plt.plot(values, probabilities, color='red')
```

```python
# Generate the sample
sample1=normal(loc=20, scale=5, size=700)
sample2=normal(loc=40, scale=5, size=300)
sample=np.hstack((sample1, sample2))
```

```python
sample
```

```python
plt.hist(sample, bins=40)
```

```python
from sklearn.neighbors import KernelDensity
model=KernelDensity(bandwidth=2.5,kernel='gaussian')
# convert the sample in 2D arrays
sample_reshaped=sample.reshape(len(sample),1)
```

```python
model.fit(sample_reshaped)
```

```python
values=np.linspace(min(sample), max(sample), 100)
values_reshaped=values.reshape(len(values),1)
```

```python
probabilities=model.score_samples(values_reshaped)
probabilities=np.exp(probabilities)
```

```python
plt.hist(sample, bins=50,density=True)
plt.plot(values_reshaped, probabilities, color='red')
plt.show()
```
