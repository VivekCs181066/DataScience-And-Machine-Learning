```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n=10 #number of trials
p=0.5 #probability of success
size=1000 #number of experiments
binomial_dist=np.random.binomial(n,p,size)
plt.hist(binomial_dist,density=True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
num_smples=10000
sample_size=300
distribution_range=(0,1)
#Generate sample form a uniform distribution
samples=np.random.uniform(distribution_range[0],distribution_range[1],size=(num_smples,sample_size))
sample_mean=np.mean(samples,axis=1)
sample_mean
```

```python
plt.hist(sample_mean,density=True,edgecolor='black',bins=30)
```
