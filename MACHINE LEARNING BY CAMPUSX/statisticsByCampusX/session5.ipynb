```python
import numpy as np
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
sns.kdeplot(df['sepal_length'],fill=True)
```

```python
temp=sorted(df['sepal_length'])
```

```python
y_quant=[]
for i in range(1,101):
    y_quant.append(np.percentile(temp,i))
```

```python
samples=np.random.normal(loc=1,scale=1,size=1000)
```

```python
x_quant=[]
for i in  range(1,101):
    x_quant.append(np.percentile(sorted(samples),i))
```

```python
sns.scatterplot(x=x_quant,y=y_quant)
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
fig=sm.qqplot(df['sepal_length'],line='s',fit=True)
plt.title("QQ plot")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Sample Quantiles")
plt.show()
```
