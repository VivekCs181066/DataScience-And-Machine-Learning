```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,Y=make_regression(n_samples=4, n_features=1,n_informative=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,Y)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(X,Y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,Y)
plt.plot(X,lr.predict(X),color='red')
```

```python
# Lets apply Gradient Descent assuming sploe is constant m=78.35
# Lets assume the starting value for intercept b=0
m=78.35
b=0
```

```python
Y_pred=(m*X + b).reshape(4)
```

```python
plt.scatter(X,Y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,Y_pred,color='green',label='GD at b=0')
plt.legend()
plt.show()
```

```python
loss_slope= -2 * np.sum(Y - m*X.ravel() - b)
loss_slope
```

```python
# Lets learning rate be 0.01
lr_rate=0.1
```

```python
step_size= lr_rate * loss_slope
```

```python
step_size
```

```python
# Calculating new intercept
b=b-step_size
```

```python
b
```

```python
Y_pred1=(m*X + b).reshape(4)
```

```python
plt.scatter(X,Y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,Y_pred1,color='green',label='GD at b={0:.2f}'.format(b))
plt.plot(X,Y_pred,color='blue',label='GD at b=0')
plt.legend()
plt.show()
```

```python
# Iteration 2
loss_slope= -2 * np.sum(Y - m*X.ravel() - b)
loss_slope
```

```python
step_size= lr_rate * loss_slope
step_size
```

```python
b=b-step_size
```

```python
b
```

```python
Y_pred2=(m*X + b).reshape(4)
```

```python
plt.scatter(X,Y)
plt.plot(X,lr.predict(X),color='red',label='OLS')
plt.plot(X,Y_pred,color='blue',label='GD at b=0')
plt.plot(X,Y_pred1,color='green',label='GD at b={0:.2f}'.format(b))
plt.plot(X,Y_pred2,color='orange',label='GD at b={0:.2f}'.format(b))
plt.legend()
plt.show()
```

```python
# Iteration 3
loss_slope= -2 * np.sum(Y - m*X.ravel() - b)
loss_slope
```

```python
step_size= lr_rate * loss_slope
step_size
```

# Putting all those code in a loop

```python
b=-100
m=78.35
loss_rate=0.1
eposchs=10
for i in range(eposchs):
    loss_slope= -2 * np.sum(Y - m*X.ravel() - b)
    step_size= loss_rate * loss_slope
    b=b-step_size
    Y_pred=(m*X + b).reshape(4)
    plt.scatter(X,Y)
    plt.plot(X,lr.predict(X),color='red',label='OLS')
    plt.plot(X,Y_pred,color='green',label='GD at b={0:.2f}'.format(b))
    plt.legend()
    plt.show()
```
